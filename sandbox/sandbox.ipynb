FRECUENCIAS:

```python
def generar_tabla_frec(df_dataFrame: pd.DataFrame):
    tamanio = (df_dataFrame).size
    fi = df_dataFrame.value_counts().sort_index()
    Fi = fi.cumsum()
    hi = (fi/tamanio)*100
    Hi = (Fi/tamanio)*100

    df_frecuencias = pd.DataFrame({
        'Frec. Absoluta': fi,
        'Frec. Absoluta Acum.': Fi,
        'Frec. Relativa': hi,
        'Frec. Relativa Acum.': Hi
    })

    round(df_frecuencias, 1)

generar_tabla_frec(df_sandbox_bd['Título profesional'])
```
